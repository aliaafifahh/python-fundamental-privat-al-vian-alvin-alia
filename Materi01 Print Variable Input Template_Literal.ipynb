## Print

Sintaks print digunakan untuk mencetak sesuatu kedalam sebuah objek

defaultnya jika objek pada print tidak didefinisikan, maka print akan mencetak di layar

### Sintaks Dasar

`print(`<isi_yang_mau_ditampilkan>`)`

```python
print("Hello World!")
```

### Parameter

Parameter didalam sintaks print ada 4 yaitu :

1. `sep`    default : " "

2. `end`   default : "\n"

3. `file`   default : sys.stdout

4. `flush`  default : False

#### sep

```python
print("Tanpa Separator :")
print("Halo," , "Selamat" , "Datang" , "Di" , "Privat" , "Al-Vian")

print("Dengan Separator :")
print("Halo," , "Selamat" , "Datang" , "Di" , "Privat" , "Al-Vian", sep="_")
```

#### end

```python
print("Tanpa Endline :")
print("Halo," , "Selamat" , "Datang" , "Di")
print("Privat" , "Al-Vian")

print("Dengan Endline :")
print("Halo," , "Selamat" , "Datang" , "Di" , end=" ")
print("Privat" , "Al-Vian")
```

## Variable

### Deklarasi Variable

#### Variable Kosong

```python
var1 = None
print(type(var1))
```

## Template Literal

### Basic

```python
nama = "Deska"
prodi = "Sains Data"

print(f"Nama\t: {nama}")
print(f"Prodi\t: {prodi}")
```

### Format Angka

```python
usia = 42.125
harga = 10000000
diskon = 0.2

print(f"Usia pohon ini {usia:.1f} Tahun")
print(f"Harga Pohon ini Rp {harga:,}")
print(f"Namun saat ini memiliki diskon {diskon:.0%}")
```

### Format Allignment

dalam f-string, terdapat format allignment untuk mengatur jumlah slot string dan mengatur allignment string (Rata Kiri, Tengah, Rata Kanan).

format allignment :

`<` Rata Kiri

`^` Tengah/Justify

`>` Rata Kanan

format : `variabel` `:` `char pengisi` `format allignment` `slot`

char pengisi biasanya adalah spasi `" "` bila tidak diisi

format allignment biasanya rata kiri jika string, rata kanan jika angka

format allignment wajib diisi bila char pengisi di set

```python
kata = "Privat Al-Vian"
```

```python
print(f"|{kata:<35}|")
```

```python
print(f"|{kata:^35}|")
```

```python
print(f"|{kata:>35}|")
```

```python
print(f"|{kata:=<35}|")
```

```python
print(f"|{kata:=^35}|")
```

```python
print(f"|{kata:=>35}|")
```
