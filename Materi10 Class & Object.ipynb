## Class

Class adalah sebuah cetak biru (blueprint) untuk membuat suatu objek. Simpelnya class ini digunakan untuk membuat type variabel baru (Khusus Python).

### Deklarasi

Sintaks :

`class` nama_class `:`

&emsp;`def __init__(self,parameter,parameter,...) :`

&emsp;&emsp;`self.var = parameter`

```python
class ruangan_pasca :
    def __init__(self,x = "Ruangan Tanpa Nama") :
        self.nama = x                   # <-- Atribut
        
    def Tampilkan_Papan_Nama(self) :    # <-- Method
        print(f"Ruangan Pada Gedung Pasca Sarjana bernama {self.nama}")
        
        
# "ruangan_pasca" bertindak sebagai nama_class
# "x" digunakan sebagai variabel untuk menampung value saat deklarasi class
# "self" digunakan untuk memberi tahu python tentang kepemilikan variabel
# "def __init__()" adalah fungsi yang berjalan ketika deklarasi class saat pertama kali
```

Method adalah fungsi yang dimiliki oleh class

Atribut adalah variable yang dimiliki oleh class

## Object

Object adalah sesuatu yang dibuat dari class yang telah ada. Jika class adalah "Blueprint", maka Object adalah hasil dari Blueprint tersebut

### Deklarasi

```python
Ruang_Pasca_01 = ruangan_pasca("Pasca 01")
Ruang = ruangan_pasca("Kamar Mandi lt 1")

# Ruang_Pasca_01 adalah variable yang menyimpan object ruangan_pasca beranama "Pasca 01"
```

### Pemanggilan Atribut atau Method

```python
print(Ruang_Pasca_01.nama)
```

```python
Ruang_Pasca_01.Tampilkan_Papan_Nama()
```

```python
Ruang.Tampilkan_Papan_Nama()
```

## Materi OOP Lanjutan

### Encapsulation (Pembedaan Hak Akses Data)

```python
class AkunBank:
    def __init__(self, nama, pin):
        self.nama = nama
        self.__pin = pin # Atribut Privat

    def cek_pin(self, x):
        return self.__pin == x
    
ab = AkunBank("Deska",1234)
```

```python
print(ab.nama)
```

```python
print(ab.__pin)
```

```python
print(ab.cek_pin(1234))
```

### Inheritance (Pewarisan)

```python
class Robot :
    def __init__(self,nama) :
        self.nama = nama
        self.hp = 100
    
    def masih_hidup(self) :
        if self.hp > 0 :
            return True
        else :
            return False
        
    def tertembak(self,damage) :
        print(f"Robot {self.nama} Tertembak, Hp Berkurang {damage}")
        self.hp -= damage
        if not(self.masih_hidup()) :
            print(f"Robot {self.nama} Telah Mati")
            
class Robot_Tempur(Robot) :
    def __init__(self,nama,atk) :
            self.nama = nama
            self.hp = 100
            self.atk = atk
    
    def menembak(self,enemy) :
        print(f"Robot {self.nama} menembak Robot {enemy.nama}, Piu Piu!!!")
        enemy.tertembak(self.atk)
```

```python
a = Robot("Andi")
b = Robot_Tempur("Budi",1000)

b.menembak(a)
```

### Polymorphism (Banyak Bentuk)

```python
class Anjing:
    def bersuara(self):
        print("Guk Guk!")

class Kucing:
    def bersuara(self):
        print("Meow!")
```

```python
for hewan in [Anjing(), Kucing()]:
    hewan.bersuara()
```
