## Fungsi Biasa

Fungsi Biasa biasanya digunakan untuk membuat set instruksi atau set kode yang sering digunakan secara berulang.

sintaks :

`def` nama_fungsi`(`parameter,parameter,...`)` `:`

&emsp;kode

`def` digunakan untuk mendefinisikan sebuah fungsi

`parameter` digunakan untuk menampung value yang akan dibawa ke fungsi, Parameter bersifat opsional.

Jika ingin membuat default value pada parameter, cukup tambahkan `=` setelah parameter lalu isi default valuenya.



```python
# Tanpa Default Value
def tampilkan_nama1(nama) :
    print(f"Halo, Nama Saya {nama}")
    
# Dengan Default Value
def tampilkan_nama2(nama = "Afim") :
    print(f"Halo, Nama Saya {nama}")
    
```

```python
tampilkan_nama1("Deska")
```

## Fungsi Return

Fungsi return biasanya digunakan untuk mencari sebuah value, sintaks nya seperti fungsi biasa, namun hanya saja tambahkan return pada akhir fungsi untuk mengembalikan sebuah value

```python
def perkalian(angka1 = 1,angka2 = 1) :
    hasil = 0
    
    for i in range(angka2) :
        hasil += angka1
        
    return hasil
```

```python
angka = perkalian(3,4)

print(angka)
```

## Fungsi Rekursif

Fungsi Rekursif adalah fungsi yang memanggil dirinya sendiri, Biasanya digunakan untuk menggantikan looping atau membuat sebuah algoritma yang memiliki siklus pengulangan fleksible.

```python
def perkalian_rekursif(angka1,angka2) :
    if angka2 == 1 :
        return angka1
    return angka1 + perkalian_rekursif(angka1,angka2-1)
```

```python
angka = perkalian_rekursif(3,4)

print(angka)
```

### Fungsi Iterable

#### `range(start,stop,step)`

range digunakan untuk membuat list instant berisi angka yang terurut

1. start = 0

```python
print(list(range(10)))
```

##### `range(start,stop)`

#### `enumerate(iterable,indeks_mulai)`

enumerate digunakan untuk mengambil nilai sekaligus indeksnya saat iterasi. Data berupa tuple per elemen (indeks,elemen).

```python
var9 = ["Mangga","Pisang","Jambu"]

print(list(enumerate(var9,0)))
```

#### `zip(iterable,iterable,...)`

zip digunakan untuk menggabungkan 2/lebih list menjadi list 2 dimensi. Cara penggabungangnya cukup sederhana, yaitu dengan cara menggabungkan elemen dengan indeks yang sama. Jumlah elemen hasil zip mengikuti jumlah elemen list terpendek

```python
nama = ["Rakha","Akhsa","Prabu","Nabil"]
kelas = ["A","B","A","A"]
nim = ["L0226025","L0226004","L0226055","L0226043"]

print(list(zip(nama,kelas,nim)))
```

#### `reversed(iterable)`

digunakan untuk membalikan iterasi

```python
nilai = [90,78,98,80,93,86,57]

print(list(reversed(nilai)))
```

#### `map(function, iterable)`

digunakan untuk menerapkan fungsi ke semua elemen iterable. Hasil dari `map` tidak langsung berbentuk struktur data, jadi diperlukan sintaks konversi struktur data seperti `list` untuk menampilkannya.

```python
var = [1.34, 54.32, 3.76, 2.712, 7.11, 45.5332]

def pembulatan(x) :
    return int(round(x,0))

print(list(map(pembulatan,var)))
```

#### `filter(function, iterable)`

digunakan untuk memfilter iterable menggunakan fungsi yang menghasilkan nilai boolean (`True`/`False`). Hasil dari `filter` tidak langsung berbentuk struktur data, jadi diperlukan sintaks konversi struktur data seperti `list` untuk menampilkannya.

```python
def kurang_dari_10(x) :
    return x < 10

print(list(filter(kurang_dari_10,var)))
```
