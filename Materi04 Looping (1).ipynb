## While Loop

while loop adalah bentuk perulangan di Python yang mengeksekusi blok kode secara berulang selama suatu kondisi bernilai `True`

### Sintaks Dasar

`while` <kondisi/boolean> `:`

&emsp; <kode_loop/>

```python
counter = -10

while counter <= 5 :
    print(f"Loop ke {counter}")
    counter += 1
    
print("Program Selesai")
```

### `break`

digunakan untuk menghentikan looping

```python
counter = 1

while counter <= 5 :
    print(f"Loop ke-{counter}")
    counter += 1
    if counter == 4 :
        break
```

### `continue`

digunakan untuk langsung lompat ke iterasi berikutnya

```python
counter = 1

while counter <= 5 :
    print(f"Loop ke-{counter}")
    counter += 1
    if counter == 4 :
        continue
    print("ABC")
```

### `while else`

selain `if`, `else` juga bisa digunakan pada `while`. Kode pada `else` akan dieksekusi jika looping `while` selesai / berhenti tanpa adanya `break`

```python
# Dengan break

counter = 1

while counter <= 5 :
    print(f"Loop ke-{counter}")
    counter += 1
    if counter == 4 :
        break
else :
    print("ABC")
    
print("Program Selesai")
```

```python
# Tanpa break

counter = 1

while counter <= 5 :
    print(f"Loop ke-{counter}")
    counter += 1
else :
    print("ABC")
    
print("Program Selesai")
```

### Infinite Loop

Infinite loop digunakan ketika ingin melakukan looping terus menerus berdasarkan input pengguna atau program

```python
while True :
    print("A")
    if input() == "0" :
        break

print("Program Selesai")
```

## For Loop

for loop adalah bentuk perulangan di Python yang mengeksekusi blok kode secara berulang menggunakan elemen elemen yang berada didalam suatu iterable, Jadi looping For akan diulang sebanyak jumlah elemen iterable yang dicantumkan

#### Sintaks Dasar

`for` variable `in` iterable `:`

Iterable adalah struktur data yang dapat menampung banyak elemen sekaligus

```python
for i in ["Pisang","Mangga","Nanas"] :
    print(i)
```

```python
buah = ["Pisang","Mangga","Nanas"]

for i in buah :
    print(i)
```

Biasanya for loop menggunakan `range()` untuk membuat iterable instant


Apa itu Range?

range digunakan untuk membuat list instant berisi angka yang terurut

sintaks dasar : `range(start,stop,step)`

default :

1. start = 0

2. step = 1

1 Parameter : `range(stop)`

2 Parameter : `range(start,stop)`

3 Parameter : `range(start,stop,step)`

sintaks `break`, `continue`, dan `else` juga bekerja pada For Loop. Cara kerjanya sama dengan yang ada di While Loop

```python
for i in range(5) :
    print("Halo",i)
```
