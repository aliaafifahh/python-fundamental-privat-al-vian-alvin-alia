## String

### Karakteristik

1. Immutable (Tidak Dapat Diubah)

Setelah string dibuat, karakternya tidak bisa diubah secara langsung di memori. Setiap operasi yang memodifikasi string akan menghasilkan objek string baru.

2. Sequence (Tipe Data Berurutan)

Karakter dalam string tersusun rapi berdasarkan urutan/indeks, sehingga bisa diakses menggunakan indeks berbasis nol (zero-based indexing).

3. Flexible Enclosure

String bisa dibuat menggunakan tanda petik tunggal `'`, petik ganda `"`, atau petik tiga `'''` / `"""` untuk multiline string.

### Indeks dan Slicing

Sama seperti `list`, `String` juga bisa diakses menggunakan indeks dan metode slicing karena pada dasarnya `String` adalah `list` yang berisi `char`

```python
kalimat = """String adalah salah satu tipe data paling mendasar di Python
yang digunakan untuk menyimpan data berupa teks, karakter, atau gabungan angka dan simbol."""

print(kalimat[:10:])
```

### Fitur Format Huruf

#### `.upper()` dan `.lower()` Memperbesar dan Mengecilkan Semua Huruf

```python
kalimat = "Aku anak FATISDA"

print(f"Menggunakan lower : {kalimat.lower()}")

print(f"Menggunakan upper : {kalimat.upper()}")
```

#### `.title()` dan `.capitalize()` Huruf besar diawal setiap kata atau kalimat

```python
kalimat = "Aku anak FATISDA"

print(f"Menggunakan title : {kalimat.title()}")

print(f"Menggunakan capitalize : {kalimat.capitalize()}")
```

#### `.swapcase()` Membalikan Kapitalisasi

```python
kalimat = "Aku anak FATISDA"

print(f"Menggunakan swapcase : {kalimat.swapcase()}")
```

### Fitur Pembersihan dan Pencarian

#### `.strip()` Pembersihan Spasi di Awal dan Akhir

```python
kalimat = "      Es Teh      "

print(f"Tanpa strip : {kalimat}")

print(f"Dengan strip : {kalimat.strip()}")
```

#### `.find(kata)` Mencari indeks Kata dalam String

```python
kalimat = "String adalah salah satu tipe data paling mendasar di Python yang digunakan untuk menyimpan data berupa teks, karakter, atau gabungan angka dan simbol."

print(kalimat.find("Python"))

print(kalimat.find("python")) # Jika tidak ada akan menghasilkan -1
```

#### `.replace(x,y)` Mengganti x menjadi y

```python
kalimat = "Distributed Big Data Processing With Spark"

kalimat = kalimat.replace("Big","Small")

print(kalimat)
```

### Fitur Pemecahan dan Penggabungan

```python
kalimat = "Distributed Big Data Processing With Spark"
```

#### `.split(separator)` Memecah String menjadi List

```python
kalimat1 = kalimat.split(" ")
print(kalimat1)
```

### Fitur Validasi

#### `.isdigit()`,`.isalpha()`,`.isalnum` Cek apakah String hanya Terdiri dari Angka, Huruf, atau Keduanya

```python
string1 = "1945"
string2 = "UNS"
string3 = "Fatisda2022"
string4 = "Sains Data UNS"
```

```python
print(string1.isdigit())
print(string2.isdigit())
print(string3.isdigit())
print(string4.isdigit())
```

```python
print(string1.isalpha())
print(string2.isalpha())
print(string3.isalpha())
print(string4.isalpha())
```

```python
print(string1.isalnum())
print(string2.isalnum())
print(string3.isalnum())
print(string4.isalnum())
```

#### `.startswith(x)`,`.endswith(x)` Mengecek awalan atau akhiran string

```python
kalimat = "Fatisda UNS"

print(kalimat.startswith("F"))
print(kalimat.endswith("S"))
```

### String Formatting / Template Literal

Digunakan ketika ingin memasukan variable kedalam String

#### f-string

gunakan f sebelum string, lalu letakkan variable menggunakan kurung kuruawal `{}`

```python
nama = "Deska"
umur = 20

formatting = f"Seorang entitas bernama {nama} memiliki umur {umur} tahun"

print(formatting)
```

```python
buah = ["Mangga", "Pisang", "Jambu"]
harga = (2,1,3)

formatting = f"Sebuah toko memiliki buah {buah} dengan perbandingan harga berturut turut {harga}"

print(formatting)
```

#### Method `.format()`

masukan kurung kuruawal `{}` didalam string untuk menentukan letak variable, lalu gunakan `.format()` berisi nama variable secara urut

```python
nama = "Deska"
umur = 20

formatting = "Seorang entitas bernama {} memiliki umur {} tahun".format(nama,umur)

print(formatting)
```

```python
buah = ["Mangga", "Pisang", "Jambu"]
harga = (2,1,3)

formatting = "Sebuah toko memiliki buah {} dengan perbandingan harga berturut turut {}".format(buah,harga)

print(formatting)
```

#### Operator `%` (C-style)

masukan %_ didalam string untuk menentukan letak variable, lalu gunakan `%` diikuti dengan tuple berisi variable yang ingin dimasukan

simbol % selalu diikuti dengan karakter, karaker bergantung dengan type variable yang ingin dimasukkan

`%d` untuk integer

`%f` untuk float

`%s` untuk lainnya

```python
nama = "Deska"
umur = 20

formatting = "Seorang entitas bernama %s memiliki umur %d tahun" % (nama,umur)

print(formatting)
```

```python
buah = ["Mangga", "Pisang", "Jambu"]
harga = (2,1,3)

formatting = "Sebuah toko memiliki buah %s dengan perbandingan harga berturut turut %s" % (buah,harga)

print(formatting)
```
