## Library Introduction

Library adalah sekumpulan modul, fungsi, dan kelas siap pakai yang telah ditulis oleh orang lain yang dapat di import dan gunakan langsung ke dalam program

### Import Library

```python
import Aritagoras
import Statistika
```

### Penggunaan Library

```python
print(Aritagoras.Pertambahan(3,6))
```

```python
data = [1,2,3,4,5,6,7,8,9]

rata2 = Statistika.meann(data)

print(rata2)
```

### Penyingkatan Nama

Agar kode tidak terlalu panjang, Library yang diimport dapat diganti namanya menggunakan sintaks `as`

```python
import Aritagoras as ar
import Statistika as stat
```

```python
print(ar.Pertambahan(3,6))
```

```python
data = [1,2,3,4,5,6,7,8,9]

rata2 = stat.meann(data)

print(rata2)
```

### Import Sub Library

Agar hanya import sebagian library aja, kita dapat menggunakan sintaks `from`

```python
from Aritagoras import Euclidean_Distance as ed
```

```python
print(ed([(0,0) , (6,8)]))
```
