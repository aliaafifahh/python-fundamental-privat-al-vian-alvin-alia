## Dictionary

Dictionary adalah sebuah struktur data didalam python dimana semua elemennya berupa pasangan key dan value

### Karakteristik

1. Key-Value Pair: 

Data disimpan berpasangan. Key bertindak sebagai indeks/label untuk mengakses Value.

2. Unique Keys: 

Key dalam satu dictionary harus unik (tidak boleh ada yang kembar). Jika Anda memasukkan key yang sama, nilai lama akan tertimpa.

3. Mutable: 

Isi dictionary (baik key maupun value) dapat ditambah, diubah, atau dihapus setelah dibuat.

4. Ordered (Terurut): 

Dictionary menjaga urutan posisi data sesuai waktu dimasukkan (insertion order).

### Manipulasi Dasar Dictionary

#### Deklarasi Dictionary

```python
dict_kosong = {}     # Dictionary Kosong

dict_berisi1 = {"Pandu" : 89, "Andri" : 95, "Tia" : 100, "Wulan" : 86, "Danis" : 91}

dict_berisi2 = {
    "Pandu" : 89,
    "Andri" : 95,
    "Tia" : 100,
    "Wulan" : 86,
    "Danis" : 91
}

```

#### Menambah & Mengubah Data

sintaks dasar : `nama_dict[key] = value`

```python
print(f"dict_kosong sebelum :\n{dict_kosong}")

dict_kosong["Deska"] = 90

print(f"dict_kosong sesudah :\n{dict_kosong}")
```

```python
print(f"dict_kosong sebelum :\n{dict_kosong}")

dict_kosong["Deska"] = 85

print(f"dict_kosong sesudah :\n{dict_kosong}")
```

#### Mengakses Data

Sama seperti list, hanya saja indeks diganti dengan key

```python
print(dict_kosong["Deska"])
print(dict_kosong["Afim"])
```

Menggunakan value default dengan `get`

```python
print(dict_kosong.get("Deska",0))
print(dict_kosong.get("Afim",0))
```

#### Menghapus Data

##### `.pop(key)` Menghapus pasangan key-value berdasarkan key

```python
print(f"Dictionary Sebelum Pop : {dict_berisi1}")

data_dihapus = dict_berisi1.pop("Andri")

print(f"Dictionary Sesudah Pop : {dict_berisi1}")

print(f"Data yang dihapus : {data_dihapus}")
```

##### `.popitem()` Menghapus pasangan key-value terakhir

```python
print(f"Dictionary Sebelum Popitem : {dict_berisi1}")

data_dihapus = dict_berisi1.popitem()

print(f"Dictionary Sesudah Popitem : {dict_berisi1}")

print(f"Data yang dihapus : {data_dihapus}")
```

##### `del` Menghapus langsung pasangan key-value dalam memory

```python
print(f"Dictionary Sebelum del : {dict_berisi1}")

del dict_berisi1["Tia"]

print(f"Dictionary Sesudah del : {dict_berisi1}")
```

##### `.clear()` Menghapus seluruh isi Dictionary

```python
print(f"Dictionary Sebelum Clear : {dict_berisi1}")

dict_berisi1.clear()

print(f"Dictionary Sesudah Clear : {dict_berisi1}")
```

#### Iterasi Pada Dictionary

##### `.keys()` Mengambil semua key dalam dictionary dalam bentuk list

```python
print(f"Dictionary : {dict_berisi2}")
hasil = dict_berisi2.keys()
print(f"Hasil : {hasil}")
```

##### `.values()` Mengambil semua value dalam dictionary dalam bentuk list

```python
print(f"Dictionary : {dict_berisi2}")
hasil = dict_berisi2.values()
print(f"Hasil : {hasil}")
```

##### `.items()` Mengambil semua pasangan key-value dalam dictionary dalam bentuk list of tuple

```python
print(f"Dictionary : {dict_berisi2}")
hasil = dict_berisi2.items()
print(f"Hasil : {hasil}")
```

#### Menggabungkan Dictionary

penggabungan dictionary dapat dilakukan dengan cara menambahkan karakter `|` diantara dictionary yang akan digabung, Jika ada key yang sama, value yang digunakan adalah dictionary terakhir

```python
dictionary1 = {
    "Wanda" : 54,
    "Vira" : 78,
    "Rara" : 46
}

dictionary2 = {
    "Vira" : 50,
    "Reyna" : 29,
    "Caca" : 68
}

print(f"Dictionary 1\t= {dictionary1}")
print(f"Dictionary 2\t= {dictionary2}\n")

gabungan = dictionary1 | dictionary2

print(f"Gabungan\t= {gabungan}")
```

## Set

Set digunakan ketika ingin membuat list tanpa value duplikat didalamnya. Cara kerjanya sama seperti dictionary namun hanya menyimpan key saja.

### Karakteristik

1. Unordered (Tidak Terurut): 

Elemen dalam set tidak memiliki posisi atau indeks tetap. Elemen dapat muncul dalam urutan berbeda setiap kali dijalankan.

2. Unindexed (Tidak Memiliki Indeks): 

Tidak bisa mengakses elemen set menggunakan set[0] atau slicing set[1:3].

3. Unique Elements (Elemen Unik): 

Set secara otomatis menghapus nilai duplikat.

4. Mutable (Set-nya): 

Elemen set dapat ditambah dan dihapus.

5. Elemen Wajib Immutable (Hashable): 

Elemen di dalam set harus berupa tipe data yang immutable (seperti `int`, `float`, `string`, atau `tuple`). List dan Dictionary tidak bisa dimasukkan ke dalam set.

### Deklarasi Set

```python
set_kosong = set()

set_berisi = {1,2,3,4,5,6}
```

### Manipulasi Elemen Set

#### `.add(`x`)` Menambah 1 Elemen

```python
print(f"Set Sebelum add : {set_berisi}")

set_berisi.add(10)

print(f"Set Setelah add : {set_berisi}")
```

#### `.update(`iterable`)` Menambah Banyak Elemen Sekaligus

```python
print(f"Set Sebelum update : {set_berisi}")

set_berisi.update([5,6,7,8])

print(f"Set Setelah update : {set_berisi}")
```

#### `.remove(`x`)` Menghapus Elemen, `KeyError` Jika elemen tidak ada

```python
print(f"Set Sebelum remove : {set_berisi}")

# set_berisi.remove(9)
set_berisi.remove(10)

print(f"Set Setelah remove : {set_berisi}")
```

#### `.discard(`x`)` Menghaspu Elemen Secara Aman

```python
print(f"Set Sebelum discard : {set_berisi}")

set_berisi.discard(8)

print(f"Set Setelah discard : {set_berisi}")
```

#### `.clear()` Mengosongkan Set

```python
print(f"Set Sebelum clear : {set_berisi}")

set_berisi.clear()

print(f"Set Setelah clear : {set_berisi}")
```

### Operasi Himpunan

#### Gabungan

menggunakan `|` atau `.union(`set`)`

```python
a = {1,2,3,4}
b = {3,4,5,6}

print("Menggunakan | :")
print(a | b)

print("Menggunakan union : ")
print(a.union(b))
```

#### Irisan

menggunakan `&` atau `.intersection(`set`)`

```python
a = {1,2,3,4}
b = {3,4,5,6}

print("Menggunakan & :")
print(a & b)

print("Menggunakan intersection : ")
print(a.intersection(b))
```

#### Selisih

menggunakan `-` atau `.difference(`set`)`

```python
a = {1,2,3,4}
b = {3,4,5,6}

print("Menggunakan - :")
print(a - b)

print("Menggunakan difference : ")
print(a.difference(b))
```

#### Beda Setara

menggunakan `^` atau `.symmetric_difference(`set`)`

```python
a = {1,2,3,4}
b = {3,4,5,6}

print("Menggunakan ^ :")
print(a ^ b)

print("Menggunakan symmetric_difference : ")
print(a.symmetric_difference(b))
```

### Method Pemeriksaan Hubungan Set

#### `.issubset(`set`)` 

Mengecek apakah seluruh elemen set merupakan bagian dari set lain

```python
a = {1,2,3,4}
b = {1,2,3,4,5,6,7}

print(a.issubset(b))
print(b.issubset(a))
```

#### `.issuperset(`set`)`

Mengecek apakah set mencakup seluruh elemen dari set lain

```python
a = {1,2,3,4}
b = {1,2,3,4,5,6,7}

print(a.issuperset(b))
print(b.issuperset(a))
```

#### `.isdisjoint(`set`)`

Mengecek apakah dua set sama sekali tidak memiliki elemen yang sama

```python
a = {1,2,3,4}
b = {1,2,3,4,5,6,7}

print(a.isdisjoint(b))
print(b.isdisjoint(a))
```
