## List

List digunakan ketika kita ingin menyimpan banyak data / value didalam 1 variable

### Karakteristik

1. Ordered (Terurut) : 

Elemen-elemen dalam list menyimpan urutan posisinya secara tetap. Elemen baru yang ditambahkan akan selalu berada di akhir.

2. Mutable (Dapat Diubah) : 

Isi dari list dapat ditambah, dihapus, atau diubah nilainya setelah list dibuat tanpa mengubah alamat memori objek list itu sendiri.

3. Heterogen : 

List dapat menampung berbagai tipe data sekaligus dalam satu variabel (string, integer, float, boolean, bahkan list lain/nested list).

4. Indexed (Berindeks) : 

Menggunakan indeks berbasis nol (zero-based indexing). Elemen pertama berada pada indeks 0, dan elemen terakhir bisa diakses dengan indeks negatif -1.

indeks adalah angka urutan setiap elemen pada list

penomoran indeks dimulai dari angka 0

sebagai contoh :

mahasiswa = ["Yuda","Falah","Alvin","Ian"]

                 /\    /\    /\   /\

                 ||    ||    ||   ||

indeks ke :&emsp;&emsp;&emsp;0&emsp;&emsp;&ensp;1&emsp;&emsp;&emsp;2&emsp;&ensp;&ensp;3

### Fitur Manipulasi Elemen

```python
var3 = ["Mangga","Pisang","Jambu"]
```

#### Cara Akses Elemen

Sintaks Dasar : `<nama_var>[indeks]`

```python
buah_diambil = var3[0]

print(buah_diambil)
```

#### Cara Edit Elemen

```python
print(var3)

var3[0] = "Anggur"

print(var3)
```

#### `.append(x)` Menambah Elemen

`.append(x)` digunakan untuk menambahkan elemen kedalam list. `.append(x)` selalu menambahkan elemen di akhir list

```python
print("Sebelum append :")
print(var3)

var3.append("Pepaya")

print("Sesudah append :")
print(var3)
```

#### `.insert(indeks,x)` Menyisipkan Elemen

`.insert(indeks,x)` digunakan untuk menambahkan elemen sesuai dengan indeks yang dimasukan

```python
print("Sebelum insert :")
print(var3)

var3.insert(2,"Leci")

print("Setelah insert :")
print(var3)
```

#### `.pop(indeks)` Menghapus Elemen

`.pop(indeks)` digunakan untuk menghapus elemen pada list, Sertakan Variabel penampung jika ingin menyimpan elemen yang dihapus. Jika indeks tidak diisi, maka program akan menghapus elemen terakhir.

```python
print("Sebelum pop :")
print(var3)

buah_diambil = var3.pop(2)

print("Setelah pop :")
print(var3)

print(f"\nBuah yang dihapus : {buah_diambil}")
```

#### `.extend(list)` Menggabungkan List

`extend(list)` digunakan untuk menggabungkan list dengan objek yang dapat menampung banyak elemen (list,tuple,set,dictionary). set dan dictionary hanya mengambil key-nya saja

```python
list_baru = ["Semangka","Pisang"]

print("Sebelum extend :")
print(var3)

var3.extend(list_baru)

print("Setelah extend :")
print(var3)
```

#### `.remove(elemen)` Menghapus Elemen yang Sama

`.remove(elemen)` Menghapus Elemen Pertama yang Sama dengan elemen yang diinput. Pengecekkan dimulai dari indeks ke-0

```python
print("Sebelum remove :")
print(var3)

var3.remove("Pisang")

print("Setelah remove :")
print(var3)
```

#### `.clear()` Mengosongkan List

`.clear()` digunakan untuk menghapus semua elemen didalam list

```python
print("Sebelum clear :")
print(var3)

var3.clear()

print("Setelah clear :")
print(var3)
```

### Fitur Pencarian & Pengurutan Data

```python
var4 = [9,2,56,5,43,6,4,5]
```

#### `.index(x)` Mencari Indeks Elemen

`.index(x)` digunakan untuk mencari indeks elemen pertama yang sama dengan elemen yang diinput.

```python
elemen_dicari = 5

print(f"Indeks elemen {elemen_dicari} adalah {var4.index(elemen_dicari)}")
```

#### `.count(x)` Menghitung jumlah elemen

`.count(x)` Menghitung jumlah elemen list yang sama dengan value yang diinput

```python
elemen_dicari = 5

print(var4)

print(f"Jumlah elemen {elemen_dicari} adalah {var4.count(elemen_dicari)}")
```

```python
A = "ABCABCABC"

print(A.count("ABCA"))
```

#### `.sort(key,reverse)` Mengurutkan elemen

`.sort(key,reverse)` digunakan untuk mengurutkan elemen list.

```python
print("Sebelum Sorting :")
print(var4)

var4.sort(reverse=True)

print("Sesudah Sorting :")
print(var4)
```

#### `.reverse()` Membalik urutan elemen

`.reverse()` digunakan untuk membalik urutan elemen didalam list.

```python
var = [5,7,13,2,1,3,20]

print("Sebelum reverse :")
print(var)

var.reverse()
print("Sesudah reverse :")
print(var)
```

### Slicing

Slicing digunakan untuk mengambil beberapa elemen sekaligus

sintaks dasar : `<nama_var>[start:stop:step]`

default :

start = 0

stop = panjang list

step = 1

```python
var5 = [1,3,5,7,9,11,13,15]

print(var5[1::3])
```

### List Multidimensi

List multidimensi adalah list yang berisi list, list yang berada didalamnya merupakan list dimensi ke 2

```python
var5 = [[1,2,3],[4,5,6],[7,8,9]]

# bisa juga dideklarasikan seperti ini

var5 = [[1,2,3],
        [4,5,6],
        [7,8,9]]
```

```python
# Cara Akses Elemen

print(var5[1][0])
```

### Dekomposisi List menggunakan Match_Case

```python
pilihan = ["A","C"]

match pilihan :
    case [pil1,pil2] if pil1 == "A" and pil2 == "B" :
        print("Case Jalan")
    case _ :
        print("Case tidak berjalan")
```

## Tuple

Simplenya tuple adalah list yang sudah paten, sekali dideklarasikan tidak dapat diubah

### Karakteristik

1. Immutable (Tidak Dapat Diubah): 

Setelah tuple dibuat, elemen di dalamnya tidak bisa ditambah, dihapus, diubah nilainya, atau diurutkan ulang.

2. Ordered (Terurut): 

Elemen dalam tuple memiliki urutan yang tetap dan diakses menggunakan indeks berbasis nol (0).

3. Heterogen: 

Dapat menampung berbagai tipe data sekaligus (integer, string, boolean, list, hingga tuple lain).

4. Sintaksis: 

Umumnya ditulis menggunakan tanda kurung biasa `()` meskipun pada praktiknya koma `,` lah yang sebenarnya mendefinisikan sebuah tuple.
